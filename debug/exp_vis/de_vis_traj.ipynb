# TensorVisHelper 2D trajectory visualization

Set the HDF5 path and tensor key below. Each loaded rollout has shape `(T_i, D)`; for `visualize2d`, `D` must be 2.

```python
from pathlib import Path

from robokit.data_manager.tensor_vis_helper import TensorVisHelper
```

```python
TASK_INSTRUCTION = "push the T-block into the green area.(dom=none)"
# TASK_INSTRUCTION = "push the T-block into the green area.(dom=none)".replace(" ", "_")

H5_FILE_PATH = Path(f"/home/geyuan/code/cospred2nvidia/output/vis_traj/{TASK_INSTRUCTION}_vis_traj.h5")
TENSOR_KEY = "actions"
OUTPUT_IMAGE_PATH = Path(f"/home/geyuan/code/cospred2nvidia/output/vis_traj/{TASK_INSTRUCTION}_vis_traj.png")

print(H5_FILE_PATH)
```

```python
loaded = TensorVisHelper.load(H5_FILE_PATH)
for key, rollouts in loaded.items():
    print(f"{key}: {[trajectory.shape for trajectory in rollouts]}")
```
